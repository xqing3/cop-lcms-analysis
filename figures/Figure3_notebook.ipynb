### Figure 3 legend

**Figure 3. Feature-based molecular network (FBMN) of heated versus unheated
cholesterol.** MS/MS spectra from all injections were assembled into a molecular
network on GNPS (FBMN job b2f1ab17) from the MZmine ion-identity (IIMN) export and
visualized with the group palette used in Figure 1. Each node is an
ion-identity-collapsed feature labeled with its MZmine feature ID; node size is
proportional to the summed precursor intensity, and the pie chart shows the
relative contribution of the heated 1000 µg/mL (teal), heated 100 µg/mL (amber),
and unheated 10 µg/mL (slate) groups. Edges connect features whose MS/MS spectra
share a modified-cosine similarity ≥ 0.70, with line width scaled to the cosine
score. Only connected clusters are shown (single, unconnected features are
omitted), and clusters are numbered (Cluster 1–Cluster 5) for reference in the
text. Nodes with a spectral-library match are outlined in black.

**Identified features.** 336 = 7α-hydroxy-4-cholesten-3-one and 488 =
Cholest-4,6-dien-3-one (cholesterol oxidation products, Cluster 1); 522 =
cholesterol (Cluster 2); 1 = a tetramethylpiperidine derivative (Cluster 4). The
COPs in Cluster 1 form a family distinct from the parent cholesterol in Cluster 2.

**Removed contaminant clusters.** Two connected clusters were identified as
non-biological background and excluded from the displayed network: (i) a
polyethylene-glycol (PEG) series of ethylene-oxide–spaced ions (features 69, 70,
73, 75, 76, 78, 80 and 193; feature 75 = library-matched
3,6,9,12-tetraoxatetracosan-1-ol), and (ii) a high-mass, regularly spaced polymer
series (features 587, 623, 651, 652, 691, 724, 725, 767, 769, 784, 786 and 810).
